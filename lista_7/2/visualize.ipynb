```python
import pandas as pd
from matplotlib import pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("results_p1.csv")
df.columns
```

```python
grouped = df.groupby(["p", "size"]).mean().reset_index().drop("i", axis=1)
grouped
```

```python
sns.lineplot(data=grouped, x="p", y="classic", hue="size")
```

```python
melted = df.drop(["i"], axis=1).melt(
    id_vars=["p","size"], var_name="kind", value_name="value"
)
```

```python
fig = sns.relplot(
    data=melted,
    x="p",
    y="value",
    col="size",
    col_wrap=2,
    hue="kind",
    kind="line",
    errorbar="ci",
    marker="o",
    legend="full",
    sizes=(0.25, 2.5),
)
fig.set(xscale="log")
fig.set(yscale="log")

plt.tight_layout()
plt.show()
```
